```python
import polars as pl
```

```python
schema = {
    "user_id": pl.String,
    "item_id": pl.String,
    "rating": pl.Float64,
    "timestamp": pl.Float64
}

df = pl.read_csv(
    "Amazon_All_Beauty.inter",
    separator="\t",
    new_columns=list(schema.keys()),
    schema=schema
)
```

```python
df.height
```

```python
df = df.filter(pl.col("rating") >= 4)

df.height
```

```python
df.describe()
```

```python
df["user_id"].shape, df["user_id"].unique().shape
```

```python
df.group_by("user_id").agg(pl.count()).describe()
```

```python
from datasets import load_dataset

url = ("https://huggingface.co/datasets/McAuley-Lab/Amazon-Reviews-2023/"
       "resolve/main/raw/review_categories/All_Beauty.jsonl")
ds = load_dataset("json", data_files=url)
```

```python
df = pl.from_pandas(ds['train'].to_pandas())

df.height, df["user_id"].shape, df["user_id"].unique().shape
```

```python
df = df.with_columns([
    pl.col("asin").alias("item_id")
]).drop([   
    "asin",
    "helpful_vote",
    "title",
    "images",
    "text",
    "parent_asin",
    "verified_purchase",
])
```

```python
df.group_by("user_id").agg(pl.count()).describe()
```
